```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import numpy as np

class Machine:
    def __init__(self, processing_times, setup_times):
        self.processing_times = processing_times
        self.setup_times = setup_times
        self.current_product = None
        self.finish_time = 0
        self.busy = False

    def process(self, product, current_time):
        self.busy = True
        setup_time = self.setup_times[self.current_product][product] if self.current_product is not None else 0
        self.current_product = product
        self.finish_time = current_time + self.processing_times[product] + setup_time
        return self.finish_time

    def update(self, current_time):
        if current_time >= self.finish_time:
            self.busy = False
            finished_product = self.current_product
            self.current_product = None
            return finished_product
        return None

class ProductionLine:
    def __init__(self, machine1_times, machine1_setups, machine2_times, machine2_setups, arrival_rate):
        self.machine1 = Machine(machine1_times, machine1_setups)
        self.machine2 = Machine(machine2_times, machine2_setups)
        self.arrival_rate = arrival_rate
        self.queue = []
        self.next_arrival = 0
        self.finished_products = [0, 0]

    def add_item(self, product, current_time):
        self.queue.append((product, current_time))

    def run_step(self, current_time):
        if self.queue and not self.machine1.busy:
            product, _ = self.queue.pop(0)
            self.machine1.process(product, current_time)

        if not self.machine2.busy and self.machine1.busy and current_time >= self.machine1.finish_time:
            finished_product = self.machine1.update(current_time)
            if finished_product is not None:
                self.machine2.process(finished_product, current_time)

        finished_product = self.machine2.update(current_time)
        if finished_product is not None:
            self.finished_products[finished_product] += 1

        if current_time >= self.next_arrival:
            self.add_item(0 if self.finished_products[0] <= self.finished_products[1] else 1, current_time)
            self.next_arrival += self.arrival_rate

def animate(i, production_line, ax1, ax2, ax3, ax4):
    production_line.run_step(i)
    
    # Clear previous frame's content
    ax1.clear()
    ax2.clear()
    ax3.clear()
    ax4.clear()
    
    # Queue visualization
    queue_colors = ['blue' if prod == 0 else 'red' for prod, _ in production_line.queue]
    ax1.bar(range(len(production_line.queue)), [1]*len(production_line.queue), color=queue_colors)
    ax1.set_title('Queue Status')
    
    # Machine 1 visualization
    if production_line.machine1.current_product is not None:
        ax2.barh([0], [1], color='blue' if production_line.machine1.current_product == 0 else 'red')
    ax2.set_xlim(0, 1)
    ax2.set_title('Machine 1 Status')
    
    # Machine 2 visualization
    if production_line.machine2.current_product is not None:
        ax3.barh([0], [1], color='blue' if production_line.machine2.current_product == 0 else 'red')
    ax3.set_xlim(0, 1)
    ax3.set_title('Machine 2 Status')
    
    # Finished products visualization
    ax4.bar([0, 1], production_line.finished_products, color=['blue', 'red'])
    ax4.set_title('Finished Products')
    ax4.set_xlabel('Product Type')
    ax4.set_ylabel('Units Finished')
    ax4.set_xticks([0, 1])
    ax4.set_xticklabels(['Product 0', 'Product 1'])

    plt.suptitle(f'Elapsed Time: {i} units')

    plt.tight_layout()

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 8))
production_line = ProductionLine({0: 5, 1: 7}, {None: {0: 0, 1: 0}, 0: {1: 2}, 1: {0: 2}}, {0: 3, 1: 4}, {None: {0: 0, 1: 0}, 0: {1: 1}, 1: {0: 1}}, 10)
ani = FuncAnimation(fig, animate, frames=np.arange(0, 200, 1), fargs=(production_line, ax1, ax2, ax3, ax4), interval=100)
plt.show()
```
