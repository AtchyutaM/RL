```python
import pandas as pd
import numpy as np
import random
import time
```

```python
import matplotlib.pyplot as plt
from IPython.display import clear_output
from matplotlib.animation import FuncAnimation
import numpy as np

class Machine:
    def __init__(self, processing_times, setup_times):
        self.processing_times = processing_times
        self.setup_times = setup_times
        self.current_product = None
        self.finish_time = 0
        self.busy = False

    def process(self, product, current_time):
        self.busy = True
        setup_time = self.setup_times[self.current_product][product] if self.current_product is not None else 0
        self.current_product = product
        self.finish_time = current_time + self.processing_times[product] + setup_time
        return self.finish_time

    def update(self, current_time):
        if current_time >= self.finish_time:
            self.busy = False
            self.current_product = None

class ProductionLine:
    def __init__(self, machine1_times, machine1_setups, machine2_times, machine2_setups, arrival_rate):
        self.machine1 = Machine(machine1_times, machine1_setups)
        self.machine2 = Machine(machine2_times, machine2_setups)
        self.arrival_rate = arrival_rate
        self.queue = []
        self.next_arrival = 0
        self.finished_products = []
        self.times = []
        self.queue_sizes = []
        self.product_in_machine1 = []
        self.product_in_machine2 = []

    def add_item(self, product, current_time):
        self.queue.append((product, current_time))

    def run_step(self, current_time):
        self.times.append(current_time)
        self.queue_sizes.append(len(self.queue))
        self.product_in_machine1.append(self.machine1.busy)
        self.product_in_machine2.append(self.machine2.busy)

        # Process queue items in Machine 1
        if self.queue and not self.machine1.busy:
            product, _ = self.queue.pop(0)
            self.machine1.process(product, current_time)

        # Transfer item from Machine 1 to Machine 2
        if not self.machine2.busy and self.machine1.busy and current_time >= self.machine1.finish_time:
            self.machine2.process(self.machine1.current_product, current_time)

        # Update machine states
        self.machine1.update(current_time)
        self.machine2.update(current_time)

        # Add new item to the queue
        if current_time >= self.next_arrival:
            self.add_item(0 if len(self.finished_products) % 2 == 0 else 1, current_time)
            self.next_arrival += self.arrival_rate

def animate(i, production_line, queue_ax, machine1_ax, machine2_ax):
    print("I am here")
    production_line.run_step(i)
    
    print(f"Time: {i}, Queue: {production_line.queue}, Machine1: {production_line.product_in_machine1[-1]}, Machine2: {production_line.product_in_machine2[-1]}")

    
    # Clear axes for fresh plot
    queue_ax.clear()
    machine1_ax.clear()
    machine2_ax.clear()
    
    # Plot the queue with color coding for product type
    product_types = [product for product, _ in production_line.queue]
    colors = ['blue' if product == 0 else 'red' for product in product_types]
    queue_ax.bar(range(len(product_types)), [1]*len(product_types), color=colors)
    queue_ax.set_title('Queue Status')
    queue_ax.set_ylim(0, 1)
    queue_ax.set_xlim(0, max(10, len(product_types)))
    
    # Indicate machine busy status
    machine1_ax.barh([0], [1], color='green' if production_line.machine1.busy else 'grey')
    machine1_ax.set_xlim(0, 1)
    machine1_ax.set_title('Machine 1 Status')
    machine1_ax.axis('off')
    
    machine2_ax.barh([0], [1], color='green' if production_line.machine2.busy else 'grey')
    machine2_ax.set_xlim(0, 1)
    machine2_ax.set_title('Machine 2 Status')
    machine2_ax.axis('off')

    plt.tight_layout()

# Define processing and setup times
machine1_times = {0: 5, 1: 7}
machine1_setups = {None: {0: 0, 1: 0}, 0: {1: 2}, 1: {0: 2}}
machine2_times = {0: 3, 1: 4}
machine2_setups = {None: {0: 0, 1: 0}, 0: {1: 1}, 1: {0: 1}}
arrival_rate = 10  # Time between arrivals

# Set up the figure and axes for the plots
fig, (queue_ax, machine1_ax, machine2_ax) = plt.subplots(3, 1, figsize=(8, 6))
production_line = ProductionLine(machine1_times, machine1_setups, machine2_times, machine2_setups, arrival_rate)

# Run the animation
ani = FuncAnimation(fig, animate, frames=np.arange(0, 2000, 1), fargs=(production_line, queue_ax, machine1_ax, machine2_ax), interval=100)
plt.show()
```
